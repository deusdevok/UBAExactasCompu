```python
# 24
def media(v):
    return v.sum() / v.size

def mediana(v):
    v_ordenado = np.sort(v)
    n = v_ordenado.size
    if n % 2 == 0:
        return (v_ordenado[n//2-1] + v_ordenado[n//2])/2
    return v_ordenado[n//2]
```

```python
v = np.array([1,2,3,8,1])
media(v)
```

```python
mediana(v)
```

# 27

$$\sigma = \sqrt{\frac{\sum_{i=1}^n (x_i - \mu)^2}{n}}$$

Las diferencias en el numerador estan elevadas al cuadrado para que la funcion sea facilmente diferenciable, y para poder medir de manera efectiva las distancias a la media, y que queden positivas

```python
def desvio_estandard(v):
    mu = media(v)
    n = v.size
    suma = sum([(x-mu)**2 for x in v])
    return np.sqrt(suma / n)
```

```python
desvio_estandard(gapminder["gdpPercap"]), np.std(gapminder["gdpPercap"])
```
