# 26

```python
moneda = np.array(["cara", "ceca"])

def proporcion_caras(n):
    sample = np.random.choice(moneda, size=n)
    return np.sum(sample == "cara") / sample.size
```

```python
proporcion_caras(10)
```

```python
proporcion_caras(100)
```

```python
proporcion_caras(1000)
```

```python
proporcion_caras(100000)
```

# 27

```python
# a) 1
sample = np.random.choice(moneda, size=10000)
np.sum(sample == "cara") / np.sum(sample == "ceca")
```

```python
# b) indeterminado
sample = np.random.choice(moneda, size=10000)
np.sum(sample == "cara") - np.sum(sample == "ceca")
```

```python
# c) 0
sample = np.random.choice(moneda, size=10000)
(np.sum(sample == "cara") - np.sum(sample == "ceca")) / sample.size
```

```python
# d) 0.5
sample = np.random.choice(moneda, size=10000)
# sample

triple_cara_luego_cara = (sample[:-4] == "cara") & (sample[1:-3] == "cara") & (sample[2:-2] == "cara") & (sample[3:-1] == "cara")
# triple_cara_luego_cara

triple_cara_luego_ceca = (sample[:-4] == "cara") & (sample[1:-3] == "cara") & (sample[2:-2] == "cara") & (sample[3:-1] == "ceca")
# triple_cara_luego_ceca

triple_cara_luego_cara.sum() / (triple_cara_luego_cara.sum() + triple_cara_luego_ceca.sum())
```
