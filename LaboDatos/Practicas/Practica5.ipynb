```python
# 15
penguins = sns.load_dataset('penguins')
penguins.head()
```

```python
cols = [c for c in penguins.columns if c != 'sex']
cols_length = [c for c in penguins.columns if 'length' in c]

(
    penguins[penguins['island'] != "Biscoe"][cols]
    .rename(columns={'species': 'especie_pinguino'})
    .groupby('especie_pinguino')[cols_length]
    .agg('mean')
    .rename(columns = lambda x: x + '_mean')
)
```
